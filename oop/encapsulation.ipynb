- in py nothing is private because if you want to make any method or attribute as private then you have to put __(double underscore) before that variable name that make that attribute private.
- and if we want to access that private variable and somehow i got access of that so for accessing that we also put __ before that variable name and try to access that but it would create a new attribute for that object because that private is stored as __name (but actual is _ClassName__PrivateAtt)
- now somehow he got know about that concept then there is a simple answer "PYTHON IS FOR ADULTS"

```python
class Atm:
    def __init__(self): 
        self.__balance = 1000 #private attribute   Actually stored as _Atm__balance
        self.__PIN = ''
        self.__menu()

    def __menu(self):
        usr_input = input("""
            press 1 to create pin
            press 2 to change pin
            press 3 to check balance
            press 4 to withdraw balance
            press 5 or anything to exit
        """)
        
        if usr_input == "1":
            self.create_pin()
        elif usr_input == "2":
            self.change_pin()
        elif usr_input == "3":
            self.check_bal()
        elif usr_input == "4":
            self.withdraw()
        else:
            print("Exiting...")

    def create_pin(self):
        pin = input("Enter the PIN to create pin: ")
        self.__PIN = pin
        print("PIN created successfully!")
        self.__menu()  

    def change_pin(self):
        usr_pin = input("Enter the old pin: ")
        if usr_pin == self.__PIN:
            self.__PIN = input("Enter the new PIN!! ")
            print("PIN changed successfully!")
        else: 
            print("Enter the correct PIN!!")
        self.__menu() 

    def check_bal(self):
        usr_pin = input("Enter the pin: ")
        if usr_pin == self.__PIN:
            print(f"Your current Balance is ${self.__balance}")
        else:
            print("Enter correct pin!")
        self.__menu()  

    def withdraw(self):
        usr = input("Enter the PIN: ")
        if usr == self.__PIN:
            amount = int(input("Enter the amount to withdraw: "))
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"Your current balance is ${self.__balance}")
            else:
                print("You don't have sufficient balance!!!!!")
        else: 
            print("Enter the correct PIN!!!")
        self.__menu() 
```
