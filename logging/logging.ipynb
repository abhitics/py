```python
import logging 
logging.basicConfig(level=logging.DEBUG,format='%(asctime)s - %(levelname)s - Filename- %(filename)s - %(lineno)s - %(message)s',filename='file.log', force=True)

def withdraw(amount,balance):
    logging.info("withdraw in progress!!")
    if(amount<balance):
        balance=balance-amount
        logging.info(f"withdraw successful ,remianing balance {balance}")
    else:
        logging.warning("Insufficient")
```

```python
withdraw(5000,1000)
```

```python
logging.basicConfig(

    level=logging.DEBUG,
    format='%(asctime)s -%(levelname)s - %(message)s - Line number: %(lineno)s',
    force=True

)
```

```python
def calculator(a,b):
    logging.info("Parameter received!")
    try:
        if a<b:
            logging.warning("a is less than b")
        else:
            res=a/b
            logging.info(f'result is {res}')
    except Exception as e:
        logging.error(f"Error {e} occured ")
```

```python
calculator(10,0)
```
