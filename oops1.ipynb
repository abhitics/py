```python
class Atm:
    def __init__(self): 
        self.balance = 0
        self.PIN = ''
        self.menu()

    def menu(self):
        usr_input = input("""
            press 1 to create pin
            press 2 to change pin
            press 3 to check balance
            press 4 to withdraw balance
            press 5 or anything to exit
        """)
        
        if usr_input == "1":
            self.create_pin()
        elif usr_input == "2":
            self.change_pin()
        elif usr_input == "3":
            self.check_bal()
        elif usr_input == "4":
            self.withdraw()
        else:
            print("Exiting...")

    def create_pin(self):
        pin = input("Enter the PIN to create pin: ")
        self.PIN = pin
        print("PIN created successfully!")
        self.menu()  

    def change_pin(self):
        usr_pin = input("Enter the old pin: ")
        if usr_pin == self.PIN:
            self.PIN = input("Enter the new PIN!! ")
            print("PIN changed successfully!")
        else: 
            print("Enter the correct PIN!!")
        self.menu() 

    def check_bal(self):
        usr_pin = input("Enter the pin: ")
        if usr_pin == self.PIN:
            print(f"Your current Balance is ${self.balance}")
        else:
            print("Enter correct pin!")
        self.menu()  

    def withdraw(self):
        usr = input("Enter the PIN: ")
        if usr == self.PIN:
            amount = int(input("Enter the amount to withdraw: "))
            if amount <= self.balance:
                self.balance -= amount
                print(f"Your current balance is ${self.balance}")
            else:
                print("You don't have sufficient balance!!!!!")
        else: 
            print("Enter the correct PIN!!!")
        self.menu() 
```

# CONSTRUCTOR 
## ->special method(inside the class) that automatically invoked  whenever class's object is created  

## WHY NEED CONTRUCTOR?->>
### ->used to write configurational code(FACTORY SETUP WIZARD) and its control would not given to the object  
### ->to guarantee that every new obj is born in valid,usable,predictable state
### ->name of constructor can't change 

# SELF 
### ->in every method self is the default parameter  
### ->in contructor avery attribute or the method is initialized with self 
### ->only objects (belonging to it's class ) has access of class's attributes and methods but two methods or attributes that belong to the same class can't use other method or attribute ,so here comes the role of self that gives authority to the same class attributes and methods to use each other whnever needed  

```python
class Fraction:
    def __init__(self,x,y): #parameterized constructor 
        self.num=x
        self.den=y

    def __str__(self): #String Representation Method not constructor 
        return '{}/{}'.format(self.num,self.den)

    def __add__(self,other): #Operator Overloading Methods not constructor
        new_num=self.num*other.den+self.den*other.num
        new_den=self.den*other.den
        return '{}/{}'.format(new_num,new_den)

    def __sub__(self,other):#Operator Overloading Methods not constructor
        new_num=self.num*other.den-self.den*other.num
        new_den=self.den*other.den
        return '{}/{}'.format(new_num,new_den)

    def __mul__(self,other):#Operator Overloading Methods not constructor
        new_num=self.num*other.num
        new_den=self.den*other.den
        return '{}/{}'.format(new_num,new_den)
    def __dec__(self): #builtin  or custom contstructors
        return self.num/self.den

        
```

```python
fr=Fraction(3,4)
fr2=Fraction(2,5)
```

```python
print(fr+fr2)
print(fr-fr2)
print(fr*fr2)
fr.__dec__()
```

### __str__  -> a human-readable, clean text description of your object

```python
#without str
class Car: 
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

my_car = Car("Tata", "Red")
print(my_car) 
# Output: <__main__.Car object at 0x7f99bcf28190> 
```

```python
#with str
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    # Defining the string representation
    def __str__(self):
        return f"A beautiful {self.color} {self.brand}"

my_car = Car("Tata", "Red")
print(my_car) 
# Output: A beautiful Red Tata
```
