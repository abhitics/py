```python
class SecurityError(Exception):
    def __init__(self,message):
        self.message=message
        print(self.message)

class Google:
    def __init__(self,name,email,password,device):
        self.name=name
        self.email=email
        self.password=password
        self.device=device
    
    def login(self,email,password,device):
        
        if email!=self.email:
            raise SecurityError("email shi dal bhai")
        elif password!=self.password:
            raise SecurityError("Incorrect Password : Nikal Nikal turant nikal!!!")
        elif device!=self.device:
            raise SecurityError("tu to gya bete!!!!")
        else :
            print("Login Successfully!!!")

obj=Google("Abhishek","abhi@gmail.com","1234","Android")
n=input("Enter your email")
p=input("Enter your password")
d=input("Enter your device ")
try:
    obj.login(n,p,d)
except SecurityError as e:
    pass
else:
    print("ho gya khatm ")
finally :
    print("Tata bye bye!!!!!")    
```

There are 2 stages where error may happen in a program

- During compilation -> Syntax Error
- During execution -> Exceptions

### Syntax Error

- Something in the program is not written according to the program grammar.
- Error is raised by the interpreter/compiler
- You can solve it by rectifying the program


### Exceptions

If things go wrong during the execution of the program(runtime). It generally happens when something unforeseen has happened.

- Exceptions are raised by python runtime
- You have to takle is on the fly

#### **Examples**

- Memory overflow
- Divide by 0 -> logical error
- Database error

```python
# catching specific exception
try:
  m=5
  f = open('sample1.txt','r')
  print(f.read())
  print(m)
  print(5/2)
  L = [1,2,3]
  L[100]
except FileNotFoundError:
  print('file not found')
except NameError:
  print('variable not defined')
except ZeroDivisionError:
  print("can't divide by 0")
except Exception as e:
  print(e)
```

```python
# finally
# else
try:
  f = open('sample1.txt','r')
except FileNotFoundError:
  print('file nai mili')
except Exception:
  print('kuch to lafda hai')
else:
  print(f.read())
finally:
  print('ye to print hoga hi')
```

```python
class MyException(Exception):
  def __init__(self,message):
    print(message)

class Bank:

  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise MyException('amount cannot be -ve')
    if self.balance < amount:
      raise MyException('paise nai hai tere paas')
    self.balance = self.balance - amount

obj = Bank(10000)
try:
  obj.withdraw(5000)
except MyException as e:
  pass
else:
  print(obj.balance)
```
