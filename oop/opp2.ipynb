### objects are mutable

```python
class Point:
    def __init__(self,x,y):
        self.x_cor=x
        self.y_cor=y
    
    def __str__(self):
        return "<{},{}>".format(self.x_cor,self.y_cor)
    
    def euclidian_dist(self,other):
        return ((self.x_cor-other.x_cor)**2 + (self.y_cor-other.y_cor)**2)**0.5

    def distance_from_origin(self):
        return self.euclidian_dist(Point(0,0))


class Line:
    def __init__(self,a,b,c):
        self.a=a
        self.b=b
        self.c=c

    def __str__(self):
        return "{}x+{}y+{}".format(self.a,self.b,self.c)

    def point_on_line(line,point):
        if line.a*point.x_cor+line.b*point.y_cor+line.c==0:
            print("Point is on the line!")
        else :
            print("nhi hai line me ")
        
    def line_intersecting(line1,line2):
        if line1.a*line2.b-line1.b*line2.a !=0:
            print("Lines are intersecting!!")
        else:
            print("Lines are not intersecting!")

    def shortest_dist(line,point):
        dis=(line.a*point.x_cor+line.b*point.y_cor+line.c)/(line.a**2+line.b**2+line.c**2)**0.2
        return dis
    
    def dist_line(line,line2):
        dis=(line.a*line2.a+line.b*line2.b+line.c*line2.c)/((line.a**2+line.b**2+line.c**2)**0.2*((line2.a**2+line2.b**2+line2.c**2)**0.2))
        return dis
    

    
```

### Reference Variables

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variable to an existing object does not create a new object

```python
class Person:
    def __init__(self,n,g):
        self.name=n
        self.gender=g

def greet(person):
    print(f"My name is {person.name} and I'm {person.gender}")

p=Person("Abhi","male")
greet(p)
p1=Person("Shri","female")
```

### Instance Variable:
- variable whose value is different for each object

```python
class Person:
    def __init__(self,name,gender):
        self.name=name
        self.gender=gender

p2=Person("Abhi","Male")
p=Person("Shri","Female")

print(p.name) #name is same in both object that holds differnt values ie, Instance Variable
print(p2.name)
```

- objects in the list 
- objects can store in list and dictionary 

```python
p1=Person("Abhi","m")
p2=Person("Shek","m")
p3=Person("Shreya","f")

l=[p1,p2,p3] #list of objetcs

for i in l:
    print(i.name,i.gender)
```
